# POS Agent Revenue Intelligence Platform

## Project Overview

Nigeria has one of the largest agency banking networks in Africa, enabling financial services in both urban and underserved communities. Thousands of POS agents process cash withdrawals, deposits, transfers, airtime purchases, bill payments, and other financial services every day.

Despite the rapid growth of agency banking, managers often struggle to answer critical business questions:

- Which agents generate the highest revenue?
- Which providers are the most profitable?
- Which locations are underperforming?
- How much revenue is lost through failed transactions?
- What operational costs reduce profitability?

This project builds a realistic analytical dataset that models daily operations of Nigerian POS agents and provides actionable business intelligence through Power BI.

# Business Problem

Agency banking providers operate thousands of terminals across multiple states.

Without centralized analytics they cannot easily identify:

- High-performing agents
- Underperforming locations
- Revenue trends
- Operational inefficiencies
- Profit leakage
- Customer transaction behaviour

Management requires a centralized dashboard capable of monitoring operational performance, financial performance, geographic expansion and profitability.

# Business Rules & Assumptions

The synthetic dataset follows realistic Nigerian agency banking operations.

Business assumptions include:

- 5 major providers
    - Moniepoint
    - Opay
    - PalmPay
    - Paga
    - FirstMonie

- 37 operating regions
    - 36 states
    - FCT

- Approximately 8,000 active POS agents

- Each agent performs between 20 and 300 transactions daily depending on location.

- Urban agents process significantly higher transaction volumes than rural agents.

- Withdrawal transactions account for the largest share of daily transactions.

- Failed transactions occur at approximately 1–4%.

- Provider commissions differ across transaction types.

- Agents incur daily operational costs including:
    - Fuel
    - Internet
    - Electricity
    - Rent
    - Maintenance

- Revenue equals Provider Commission.

- Net Profit equals Revenue minus Operational Cost.

# Data Model

Fact Tables

- Fact_Transactions
- Fact_Daily_Costs

Dimension Tables

- Dim_Agent
- Dim_Date
- Dim_Provider
- Dim_Transaction_Type
- Dim_Location
